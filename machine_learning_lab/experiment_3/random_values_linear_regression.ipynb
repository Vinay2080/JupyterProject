# Random Values: Simple Linear Regression Demo

Objective: demonstrate the workflow of **simple linear regression** using randomly generated data (fit → predict → evaluate → plot).  \
Note: because `X` and `y` are random, the regression fit can be poor and **R² may be near 0**.


```python
import matplotlib.pyplot as plt
import numpy as np
```

### Generate random data
`X` is the feature (input) and `y` is the target (output). We generate 25 random integer points between 0 and 99.


```python
X = np.random.randint(100, size=25)
y = np.random.randint(100, size=25)
```

### Visualize the data
Scatter plot to see whether a linear relationship exists between `X` and `y`.


```python
plt.scatter(X, y)
plt.xlabel('X')
plt.ylabel('y')
```

### Train the model
Scikit-learn expects `X` to be **2D** (shape: `(n_samples, n_features)`), so we reshape it to `(-1, 1)`.


```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()
model.fit(X.reshape(-1, 1), y)
```

### Model parameters
- `coef_` (slope): how much `y` changes for a 1-unit increase in `X`
- `intercept_`: predicted `y` when `X = 0`


```python
model.coef_
```

```python
model.intercept_
```

### Make predictions
Compute predicted values `y_pred` for each `X`.


```python
y_pred = model.predict(X.reshape(-1, 1))
```

### Evaluate the model
- **MSE**: average squared error (penalizes large errors heavily)
- **MAE**: average absolute error (easier to interpret)
- **R²**: proportion of variance explained by the model (often low for random data)


```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
```

```python
mean_squared_error(y, y_pred)
```

```python
mean_absolute_error(y, y_pred)
```

```python
r2_score(y, y_pred)
```

### Plot the regression line
Plot the original scatter points (`X`, `y`) and the model predictions as a line (`X`, `y_pred`).  \
Tip: a cleaner line can be drawn by sorting `X` before plotting the line.


```python
plt.scatter(X, y)
plt.plot(X, y_pred, color='red')
plt.xlabel('X')
plt.ylabel('y')
```
